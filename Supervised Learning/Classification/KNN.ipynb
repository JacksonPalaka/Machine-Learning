```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# Dataset
data = {
    'Weight': [130, 140, 150, 170, 180, 200],
    'Shade':  [2,   3,   4,   7,   8,   9],   # 1 = apple-like, 10 = orange-like
    'Fruit':  ['Apple', 'Apple', 'Apple', 'Orange', 'Orange', 'Orange']
}

df = pd.DataFrame(data)

X = df[['Weight', 'Shade']]
y = df['Fruit']

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)
```

```python
model.classes_
```

```python
weight = float(input("Enter fruit weight (grams): "))
shade = int(input("Enter shade : "))

prediction = model.predict(pd.DataFrame([[weight, shade]],columns=['Weight','Shade']))[0]

print("Predicted fruit:", prediction)
```
